# TEM data → NOMAD (emingest)

Add TEM data of any kind and format (images, diffraction patterns, spectra, focal and tilt series, 4D-STEM data sets,
spectrum images …), describe them with the metadata NXem / pynxtools asks for (plus a rich-text description and the
history of the sample), import an experiment or a resource from **elabFTW**, and upload everything to **NOMAD**
(central NOMAD, test or an Oasis) — optionally converted to NeXus NXem.

* tab 1: local files or folders (in Jupyter), or files from your computer (Voilà / Solara)
* tab 2: metadata (`*` required, `°` recommended by NXem)
* tab 3: elabFTW (address of the entry + API key)
* tab 4: NOMAD API token (NOMAD → user menu → *API tokens*), target upload and folder, NXem conversion

Install: `pip install "emingest[formats] @ git+https://github.com/christophys/EMingest"`

````python
from emingest.app import show

app = show(server_mode=False)          # server_mode=True: transfer from the browser instead of local paths
````

Headless use (e.g. in a script):

```python
from emingest import convert, metadata, eln
m = metadata.empty(); m["entry"]["start_time"] = "2026-11-20 10:00"; ...
metadata.write_eln_data(m, "session.eln_data.yaml")
convert.convert_all(["a.dm4", "b.emd", "c.hspy"], "session.eln_data.yaml", "nxem/")
```
